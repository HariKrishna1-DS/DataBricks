```python
%sql

--Step1 Create external location
CREATE EXTERNAL LOCATION IF NOT EXISTS `external-data` 
-- Location Name: `external-data` 
--container Name: dbfscontainer
--STorage Lake Name: mystoragelakeadb10am
--Folder Name: external
-- URL : "abfss://dbfscontainer@mystoragelakeadb10am.dfs.core.windows.net/external"
URL "abfss://dbfscontainer@mystoragelakeadb10am.dfs.core.windows.net/external"
WITH (CREDENTIAL `adb10am-storage-credential`)
```

```python
%fs ls abfss://dbfscontainer@mystoragelakeadb10am.dfs.core.windows.net/external
```

```python
#step 1: Goto dbfscontainer->jsondata-->upload a file-->flight-time.json
#-----------------------------------------------------------------------------
base_dir="abfss://dbfscontainer@mystoragelakeadb10am.dfs.core.windows.net/"


#Step 2: providing schema
flight_schema='''FL_DATE DATE,
                OP_CARRIER STRING,
                OP_CARRIER_FL_NUM INT,
                ORIGIN STRING,
                ORIGIN_CITY_NAME STRING,
                DEST STRING,DEST_CITY_NAME STRING,
                CRS_DEP_TIME INT,DEP_TIME INT,
                WHEELS_ON INT,
                TAXI_IN INT,
                CRS_ARR_TIME INT,ARR_TIME INT,
                CANCELLED INT,
                DISTANCE INT'''

#Step 3: Creating dataframe
flight_time_df=(spark.read.format("json")
                .schema(flight_schema)
                .option("dateformat","m/d/y")
                .load(base_dir+"jsondata/flight-time.json"))
flight_time_df.show()  
```

```python
#Step.4 Saving Data In Delta Format in Extrenal Location
flight_time_df.write.format("delta").mode("overwrite").save(base_dir+"external/flight-time")
flight_time_df.write.format("delta").mode("overwrite").save("abfss://dbfscontainer@mystoragelakeadb10am.dfs.core.windows.net/external/flight-time1")
```

```python
%fs ls abfss://dbfscontainer@mystoragelakeadb10am.dfs.core.windows.net/external/flight-time
```

```python
flight_time_df.coalesce(1).write.format("delta").mode("overwrite").save("abfss://dbfscontainer@mystoragelakeadb10am.dfs.core.windows.net/external/flight-time1")
```
