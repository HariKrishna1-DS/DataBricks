```python
base_dir="abfss://dbfscontainer@mystoragelakeadb10am.dfs.core.windows.net/"
df=spark.read.json(base_dir+"jsondata/emps.json")
df.show()
```

```python
df.display()
```

```python
display(df)
```

```python
#1)select(): To selecting  a particular column
df1=df.select("ename","sal")
df1.show()
```

```python
df.printSchema()
```

```python
#2)Transformations
df2=df.select(df.ename,(df.sal+5000).alias("sal"))
df2.show()
```

```python
from pyspark.sql.functions import col
df2=df.select(col("ename"),(col("sal")+5000).alias("sal"))
df2.show()

```

```python
#3.selectExpr():
df.selectExpr("ename","sal+5000").show()
```

```python
mydf=df
mydf.show()
```

```python
#4.filter():For filtering data
# I want those emps whose salary is greater than 20000
df4=df.filter(df.sal>20000)
df4.show()
```

```python
#Filter ,select and show
df.filter(df.sal>20000).select(df.ename,df.sal).show()


```

```python
#I want employees those department no.11
df.filter(df.dno==11).select(df.ename,df.sal,df.dno).show()
```
